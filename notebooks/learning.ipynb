Black-Scholes assumes volatiity is constant, prices move smoothly with no jumpbs, and returns are perfectly lognormal.

So I will use it in this project as a translation...As Black Scholes converts prices into implied volatility based on prices.

At-the-Money (ATM): Underlying price equals the strike price (\(S = K\)).

In-the-Money (ITM): Underlying price is above the strike price (\(S > K\)). Exercising gives instant profit.

Out-of-the-Money (OTM): Underlying price is below the strike price (\(S < K\)). Exercising loses money.
